## 1. Kontrola rozměrů ★

```python
diameters = [25.01, 24.97, 25.06, 25.00, 24.94, 25.03, 25.05, 24.95, 25.10, 24.99]
target_diameter = 25
tolerance = 0.05

total_number = len(diameters)
n = 0

for diametr in diameters:
    odchylka = round(diametr - target_diameter, 3)

    if odchylka > tolerance:
        print(f"{diametr} - VELKÝ")
        n += 1
    elif odchylka < -tolerance:
        print(f"{diametr} - MALÝ")
        n += 1
    else:
        print(f"{diametr} - OK")

    
print(f"Zmetků: {n} z {total_number} ({n/total_number*100} %)")
```

## 2. Přestupné roky ★

```python
year = int(input("Zadej rok: "))


if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} je přestupný")
else:
    print(f"{year} není přestupný")


print("\n--- Výpis přestupných let ---")
year_start = 1890
year_end = 2030


for current_year in range(year_start, year_end + 1):
    if (current_year % 4 == 0 and current_year % 100 != 0) or (current_year % 400 == 0):
        print(f"{current_year} je přestupný")
```

## 3. Prvočísla ★★

```python

n = int(input("Zadek číslo: !"))
m = 1

for current_number in range(m, 10000+1):
    if current_number <= 1:
        prime = False
        print(f"{current_number} = prime: {prime}")
        continue
    if current_number <= 3:
        prime = True
        print(f"{current_number} = prime: {prime}")
        continue
    if current_number % 2 == 0 or current_number % 3 == 0:
        prime = False
        print(f"{current_number} = prime: {prime}")
        continue

    limit = int(current_number ** 0.5)

    i = 6

    while limit >= i:
        if current_number % i == 0 or current_number % (i + 2) == 0:
            prime = False
            print(f"{current_number} = prime: {prime}")
            continue
        i = i + 6 
        prime = True
    print(f"{current_number} = prime: {prime}")


print("\n-------- Začátej tvého čísla ---------")

main_n = n 
prvocinitel = []

if n <= 1:
    prime = False
    print(f"{n} = prime: {prime}")
    

limit = int(n ** 0.5)


for i in range(2, limit + 1):
    if n % i == 0:
        prime = False
        break
    else:
        prime = True
print(f"{n} = prime: {prime}")


while n % 2 == 0:
    prvocinitel.append(2)
    n = n // 2


limit = int(n ** 0.5)
for i in range(3 , limit + 1, 2):
    if n % i == 0:
        prvocinitel.append(i)
        n = n // i

if n > 2:
    prvocinitel.append(n)

text_prvocinitele = "*".join(map(str, prvocinitel))
print(f"number: {main_n} = {text_prvocinitele}")
```

## 4. Kosočtverec ★★

```python
z = False
while z == False:
    n = abs(int(input("Zadej celé liché číslo")))
    if n and n % 2 == 0:
        z = False
    else:
        z = True



for i in range (1, n + 1, 2):
    space = (n - i) // 2
    print(f"{space * " "}{i*"*"}") 

for x in range(n - 2, -1, -2):
    space = (n - x) // 2
    print(f"{space * " "}{x*"*"}") 
```

## 5. Ciferné hrátky ★★

```python
import numpy as np

n = int(input("How many digits? "))

numbers = np.arange(10 ** (n - 1), 10 ** (n))





for number in numbers:
    str_number = (str(number))
    soucet = 0
    count = n
    while count > n:
        soucet += int(number)[str_number -1] ** n
        str_number = str_number -1
        if soucet == number:
            print(f"{number}")    
  
    
```

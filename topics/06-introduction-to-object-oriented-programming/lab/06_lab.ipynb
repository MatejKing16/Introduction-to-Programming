## 1. Obdélník ★

Napište třídu `Rectangle` s atributy `width` a `height` a metodami:

- `area()` a `perimeter()` — obsah a obvod,
- `is_square()` — vrátí `True`, pokud je obdélník čtverec,
- `scale(factor)` — zvětší oba rozměry `factor`krát,
- `describe()` — vrátí řetězec, například `Obdélník 3 × 4, obsah 12`.

Vytvořte seznam alespoň tří obdélníků, vypište jejich popis a najděte
ten s největším obsahem.

````python
import math


class Rectangle():
    def __init__(self, width, height):
        self.width = width
        self.height = height
        

    def area(self):
        return self.width*self.height

    def perimetr(self):
        return 2*self.width+2*self.height


    def is_square(self):
        return self.width == self.height

    def scale(self, faktorial):
        self.width = math.factorial(faktorial)*self.width
        self.height = math.factorial(faktorial)*self.height

    def __str__(self):
        area = self.area()
        return f"Obdelník {self.width} x {self.height}, obsah {area}"



r1 = Rectangle(3,4)
r2 = Rectangle(3,5)
r3 = Rectangle(8,3)

r_list = [r1, r2, r3]


for r in r_list:
    max_area = 0
    area = r.area()
    if area > max_area:
        max_area = area

for r in r_list:
    print(r)



print(f"Nejvetší obdelník má obsah {max_area}")

n = 5
for i in range (1,n+1):
    r2.scale(i)
    print(r2)

````

Proč `add` vrací nový vektor, místo aby změnil ten stávající?

**Rozšíření:** Doplňte speciální metody `__add__`, `__mul__` a `__eq__`,
aby fungovaly zápisy `u + v`, `u * 2` a `u == v`.


````python
import math


class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y


    def length(self):
        return (self.x**2 + self.y **2)**0.5


    def add(self, other):
        xn = self.x + other.x
        yn = self.y + other.y
        return xn, yn

    def angle_to(self, other):
        return round(math.acos((self.x*other.x+self.y*other.y)/(self.length()*other.length()))*(180/math.pi))

    def __str__(self):
        return f"{self.x, self.y}"



u = Vector2D(3, 4)
v = Vector2D(1, 0)


print(u, u.length())                    # (3, 4) 5.0
print(u.add(v))                         # (4, 4)
print(v.angle_to(Vector2D(0,1)))        # (90)
````

Napište krátký scénář, který sklad naplní, několikrát z něj vydá zboží
(i neúspěšně) a vypíše závěrečný přehled.

**Rozšíření:** Sklad si vede historii všech pohybů (příjem a výdej, datum,
množství) a umí vypsat historii jedné položky.


````python
import math

class Fraction():
    def __init__(self, citatel, jmenovatel):
        if jmenovatel == 0: 
            return "cant divade by 0"

        if jmenovatel < 0 and citatel > 0:
            jmenovatel = -1*(jmenovatel)
            citatel = -1*citatel

        if jmenovatel < 0 and citatel < 0:
            jmenovatel = -1*(jmenovatel)
            citatel = -1*citatel
        
        while True:
            zlomek_gcd = (math.gcd(citatel, jmenovatel))

            if zlomek_gcd == None or zlomek_gcd == 1:
                self.jmenovatel = jmenovatel
                self.citatel = citatel
                break
            jmenovatel = jmenovatel//zlomek_gcd
            citatel = citatel//zlomek_gcd

            
        



    def __add__(self, other):
        jme_self = self.jmenovatel
        jme_other = other.jmenovatel
        cit_self = self.citatel
        cit_other = other.citatel


        zlomek_lcm = math.lcm(jme_self, jme_other)
        print(zlomek_lcm)


        cit_other = (zlomek_lcm//jme_other)*cit_other
        cit_self = (zlomek_lcm//jme_self)*cit_self

        jme_self = zlomek_lcm
        jme_other = zlomek_lcm
        
        citatel_add = cit_other+cit_self
        jmenovatel_add = jme_self
        return citatel_add, jmenovatel_add


    def __sub__(self, other):
       
        jme_self = self.jmenovatel
        jme_other = other.jmenovatel
        cit_self = self.citatel
        cit_other = other.citatel


        zlomek_lcm = math.lcm(jme_self, jme_other)
        print(zlomek_lcm)


        cit_other = (zlomek_lcm//jme_other)*cit_other
        cit_self = (zlomek_lcm//jme_self)*cit_self

        jme_self = zlomek_lcm
        jme_other = zlomek_lcm
        
        citatel_sub = cit_self-cit_other
        jmenovatel_sub = jme_self

        return citatel_sub, jmenovatel_sub


    def __mul__(self, other):
        citatel_mul = self.citatel*other.citatel
        jmenovatel_mul = self.jmenovatel*other.jmenovatel
        return Fraction(citatel_mul, jmenovatel_mul)


    
    def __str__(self):
        return f"{self.citatel}/{self.jmenovatel}"

        



a = Fraction(2,5)
b = Fraction(4, 156)
print(a, b)
print(a + b)
print(a - b)

print(Fraction(-2, 4)) 
print(a*b)


````

## 1. Hmotnost trubky ★

Napište funkce:

- `circle_area(d)` — obsah kruhu o průměru `d`,
- `tube_cross_section(d_out, d_in)` — obsah průřezu trubky
  (využijte `circle_area`),
- `tube_mass(d_out, d_in, length, density=7850)` — hmotnost trubky v kg.
  Rozměry jsou v metrech, hustota v kg/m³ a výchozí hodnota odpovídá oceli.

````python
def tube_mass(d_out, d_in, lenght, density=7850):
    s1 = (3.14 * d_out**2)/4
    s2 = (3.14 * d_in**2)/4
    sc = s1 - s2
    return round(sc*lenght*density,2)


print (tube_mass(0.06,0.050,6), "Kg")
print (tube_mass(0.06,0.050,6,density=2700), "Kg")



    
````

## 2. Převod teplot ★

Napište funkci `convert_temperature(value, from_unit, to_unit)`, kde
jednotky jsou `"C"`, `"F"` nebo `"K"`. Pro neznámou jednotku funkce vrátí
`None`. Nápověda: nejjednodušší je převést hodnotu nejdřív na °C a teprve
potom do cílové jednotky.

```python
print(convert_temperature(100, "C", "F"))   # 212.0
print(convert_temperature(0, "K", "C"))     # -273.15
print(convert_temperature(-40, "C", "F"))   # -40.0
print(convert_temperature(20, "C", "X"))    # None
```

Pomocí této funkce pak vypište převodní tabulku °C → °F → K pro teploty
od −40 do 100 °C po 20 stupních

````python
def convert_temperature(value, from_unit, to_unit):
  
    from_unit = from_unit.upper()
    to_unit = to_unit.upper()


    if from_unit == to_unit:
        return float(value)

   
    if from_unit == "F":
        celsius = (value - 32) * 5/9  # Opraven vzorec (původně bylo *1.8)
    elif from_unit == "K":
        celsius = value - 273.15
    elif from_unit == "C":
        celsius = value
    else:
        raise ValueError(f"Neznámá výchozí měrná jednotka: '{from_unit}'")


    if to_unit == "F":
        return (celsius * 9/5) + 32
    elif to_unit == "K":
        return celsius + 273.15
    elif to_unit == "C":
        return celsius
    else:
        raise ValueError(f"Neznámá cílová měrná jednotka: '{to_unit}'")

print(convert_temperature(100, "C", "F"))   # 212.0
print(convert_temperature(0, "K", "C"))     # -273.15
print(convert_temperature(-40, "C", "F"))   # -40.0

````

## 3. Statistika měření ★★

Napište funkce pro seznam naměřených hodnot:

- `mean(values)` — aritmetický průměr,
- `median(values)` — medián,
- `std_dev(values)` — směrodatná odchylka (√(součet (x − průměr)² / n)),
- `min_max(values)` — vrátí **dvojici** (nejmenší, největší) bez použití
  `min()` a `max()`,
- `summary(values)` — vrátí slovník se všemi výsledky výše.

```python
data = [2, 4, 4, 4, 5, 5, 7, 9]
print(summary(data))
# {'mean': 5.0, 'median': 4.5, 'std_dev': 2.0, 'min': 2, 'max': 9}
```

Pozor: funkce `median` potřebuje hodnoty seřadit, ale **nesmí změnit**
seznam, který dostala. Ověřte, že `data` je po zavolání `summary(data)`
stále ve stejném pořadí. Jaký je rozdíl mezi `values.sort()`
a `sorted(values)`?


````python
data = [2, 4, 4, 4, 5, 5, 7, 9]
import statistics

def mean(values):
    return statistics.mean(values)

def median(values):
    statistics.median(values)

def std_dev(values):
    Xc = mean(values)
    od = 0
    n = 0
    for value in values:
        n += 1
        od += (value - Xc)**2
    return (od/(n-1))**0.5

def min_max(values):
    max_value = min_value = values[0]
    for value in values:
        if value > max_value:
            max_value = value
        if value < min_value:
            min_value = value
    return (max_value, min_value)

def summary(values):
    print(f"""
        Průměr je: {mean(values)}
        Median je: {median(values)}
        Směrodatná odchylka je: {std_dev(values)}
        největší a nejmenší hodnoty jsou: {min_max(values)}
    
    """)
    return "Vše proběhlo uspěšně"


print(summary(data))
````

## 4. Kontrola rodného čísla ★★

Pro rodná čísla ve tvaru `RRMMDD/XXXX` platí (zjednodušeně) tato pravidla:

- celé desetimístné číslo (bez lomítka) je dělitelné 11,
- u žen je k měsíci připočteno 50,
- dvojčíslí roku `00`–`53` znamená roky 2000–2053, `54`–`99` roky 1954–1999.

Napište funkce:

- `normalize(birth_number)` — odstraní lomítko a mezery, vrátí řetězec
  10 číslic, nebo `None`, pokud vstup nemá správný tvar,
- `is_valid(birth_number)` — ověří tvar, dělitelnost 11 a platnost měsíce,
- `gender(birth_number)` — vrátí `"muž"` nebo `"žena"`,
- `birth_date(birth_number)` — vrátí datum narození jako řetězec.

```
736028/5163 → platné, žena, 28. 10. 1973
990101/1230 → platné, muž, 1. 1. 1999
736028/5164 → neplatné
```

**Rozšíření:** Ověřte, že datum opravdu existuje (nepřipustí 30. února),
včetně přestupných roků. Kolik funkcí z předchozích úloh můžete znovu
použít?


````python
def normalize(birth_number):
    if not isinstance(birth_number, str):
        return None

    # Odstraní mezery
    s = birth_number.replace(" ", "")

    # Pokud má lomítko na 7. pozici (index 6), odstraní ho
    if len(s) == 11 and s[6] == "/":
        s = s.replace("/", "")

    # Musí mít přesně 10 číslic
    if len(s) == 10 and s.isdigit():
        return s
    return None


def is_valid(birth_number):
    clean = normalize(birth_number)
    if clean is None:
        return False

    # 1. Dělitelnost 11
    if int(clean) % 11 != 0:
        return False

    # 2. Platnost měsíce (pro muže 1-12, pro ženy 51-62)
    month = int(clean[2:4])
    if not ((1 <= month <= 12) or (51 <= month <= 62)):
        return False

    return True


def gender(birth_number):
    if not is_valid(birth_number):
        return None

    clean = normalize(birth_number)
    month = int(clean[2:4])

    if 1 <= month <= 12:
        return "muž"
    else:
        return "žena"


def birth_date(birth_number):
    if not is_valid(birth_number):
        return None

    clean = normalize(birth_number)

    # 1. Rok (00–53 -> 2000–2053, 54–99 -> 1954–1999)
    raw_year = int(clean[0:2])
    year = 2000 + raw_year if raw_year <= 53 else 1900 + raw_year

    # 2. Měsíc (ženám odečteme 50)
    month = int(clean[2:4])
    if month > 50:
        month -= 50

    # 3. Den
    day = int(clean[4:6])

    # Vrátí formát např. "28. 10. 1973"
    return f"{day}. {month}. {year}"


# --- Testy podle zadání ---
print(birth_date("736028/5163"))  # 28. 10. 1973
print(gender("736028/5163"))      # žena
print(is_valid("736028/5163"))    # True

print(birth_date("990101/1230"))  # 1. 1. 1999
print(gender("990101/1230"))      # muž
print(is_valid("990101/1230"))    # True

print(is_valid("736028/5164"))    # False
print(birth_date("736028/5164"))  # None
````

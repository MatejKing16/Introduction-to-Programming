```python
t = int(input("Zadej čas v sekudách: "))


h = t // 3600
min = (t % 3600) // 60
s = t % 60


if h > 24:
    h = h % 24
    d = t // 86400
    print(f"{d} d {h}:{min:02d}:{s:02d}")
else:
    print(f"{h}:{min:02d}:{s:02d}") 
```

Rozbor slova ★

```python
slovo = str(input("Zadej slovo: "))

first_letter = slovo[0]
last_letter = slovo[-1]

number_of_letters = len(slovo)
reverse_word = slovo[::-1]

if  slovo.upper() == reverse_word.upper():
    palindrome = "true"
else:
    palindrome = "false"

print(f"""
Slovo {slovo}
První znak: {first_letter}
Poslední znak: {last_letter}
Počet znaků: {number_of_letters}
Zpětné slovo: {reverse_word.upper()}
Palindrom: {palindrome}
""")
```

4. Rozsahy celých čísel ★★

```python
bite = 8

while bite <= 64:
    range = (2 ** bite) -1
    range_negative = -1 * ((2 ** bite)/2)

    print(f"Pro počet bitů: {bite}, rozsah: {bite - bite} až {range}, nebo {range_negative} až{range/2}")
    bite = bite * 2
```

## 6. Znaky a bajty ★★

```python
text = str(input("Zadej text: "))
x = len(text)

for i in range (x):
    letter = text[i]
    type = f"U+{ord(letter):04X}"
    size = len(letter.encode("utf-8"))


    print(f"{letter}    {type}  {size} B")
```

## 7. Četnost písmen ★★★

```python
veta = input("Zadej větu: ")
veta_sorted = sorted(veta)
delka = len(veta_sorted)

i = 0

while i < delka:
    letter = veta_sorted[i]

    # 1. Kontrola, jestli to je písmeno (a nesmíme zapomenout posunout 'i')
    if not letter.isalpha():
        i += 1
        continue
    
    n = 0
    
    # 2. Počítáme stejná písmena (dokud nevyjedeme z délky textu)
    while (i + n) < delka and veta_sorted[i + n] == letter:
        n += 1
            
    # 3. Vypíšeme výsledek rovnou, jakmile cyklus najde něco jiného
    print(f"{letter}: {n}")
    
    # 4. Posuneme hlavní index o tolik písmen, kolik jsme právě napočítali
    i += n
```
